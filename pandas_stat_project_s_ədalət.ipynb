### 2.8 Outlier detection (IQR method)

Applying the IQR method to `days_since_prior_order` to formally identify outliers, complementing the visual boxplot check in section [D].

```python
q1 = orders['days_since_prior_order'].quantile(0.25)
q3 = orders['days_since_prior_order'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = orders[
    (orders['days_since_prior_order'] < lower_bound) | (orders['days_since_prior_order'] > upper_bound)
]

n_valid = orders['days_since_prior_order'].notna().sum()
print(f"Q1: {q1}")
print(f"Q3: {q3}")
print(f"IQR: {iqr}")
print(f"Lower bound: {lower_bound}")
print(f"Upper bound: {upper_bound}")
print(f"Number of outliers: {len(outliers)} ({len(outliers) / n_valid * 100:.2f}% of non-missing values)")
```

`days_since_prior_order` is capped by Instacart at 30 days (as already seen in the histogram above, which shows a spike at 30), so the IQR bounds computed here should be read with that ceiling in mind — any values flagged above the upper bound are really just the natural clustering at the 30-day cap rather than true data-entry anomalies. Given this, the values are kept as-is (not removed or clipped further); the cap is a property of how Instacart recorded the data, not a cleaning problem.

### 2.9 Feature engineering

Three new columns are added to `orders`, each aimed at a specific analytical question used later in Step 3.

```python
# Weekend flag: Saturday (6) and Sunday (0), per the day-of-week convention established in 2.1
orders['is_weekend'] = orders['order_dow'].isin([0, 6])

# Time-of-day bucket, for a coarser view than raw hour-of-day
def bucket_hour(hour):
    if 0 <= hour <= 5:
        return 'Night'
    elif 6 <= hour <= 11:
        return 'Morning'
    elif 12 <= hour <= 17:
        return 'Afternoon'
    else:
        return 'Evening'

orders['time_of_day'] = orders['order_hour_of_day'].apply(bucket_hour)

# Recency segment based on days_since_prior_order, reusing the missing-value logic from 2.6
def recency_group(days):
    if pd.isna(days):
        return 'First order'
    elif days < 7:
        return 'Frequent (<7 days)'
    elif days <= 14:
        return 'Regular (7-14 days)'
    else:
        return 'Occasional (>14 days)'

orders['order_recency_group'] = orders['days_since_prior_order'].apply(recency_group)

orders[['order_dow', 'order_hour_of_day', 'days_since_prior_order',
        'is_weekend', 'time_of_day', 'order_recency_group']].head()
```

`is_weekend` supports a direct weekday-vs-weekend comparison beyond the single-day breakdowns in [A]/[B]. `time_of_day` collapses 24 hourly buckets into four interpretable segments, used for the pie chart in [E]. `order_recency_group` turns the continuous `days_since_prior_order` into a three-tier framing directly usable in `groupby()`, instead of everyone re-deriving the same buckets ad hoc.

### 2.10 Data type correction

Columns that are categorical in nature are cast to pandas' `category` dtype, which is more memory-efficient and signals intent for downstream grouping/plotting. `order_dow` is intentionally **kept as `int`**, even though it is conceptually categorical/ordinal, because it is repeatedly used in numeric-style comparisons and `.isin()` filters throughout the rest of the notebook (e.g. section [B]); converting it here would risk breaking that already-written code. No table in this dataset contains a real calendar date (only relative day-of-week and hour-of-day), so no `to_datetime` conversion applies.

```python
orders['time_of_day'] = orders['time_of_day'].astype('category')
orders['order_recency_group'] = orders['order_recency_group'].astype('category')
departments['department'] = departments['department'].astype('category')
aisles['aisle'] = aisles['aisle'].astype('category')

orders[['time_of_day', 'order_recency_group']].dtypes
```

## Dashboard

```python
import ipywidgets as widgets
from IPython.display import display

# ---------- Precompute a single merged frame once, reused by every slicer update ----------
# One row per (order, product) line, already carrying department + timing info,
# so the slicer callback below never has to re-merge the 4.5M-row order_products table.
op_full = (
    order_products
    .merge(products[['product_id', 'product_name', 'department_id']], on='product_id', how='left')
    .merge(departments, on='department_id', how='left')
    .merge(orders[['order_id', 'order_dow', 'order_hour_of_day']], on='order_id', how='left')
)

DOW_LABELS = ['Sun', 'Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat']  # 0 = Sunday, per project note
HOUR_LABELS = [f"{h:02d}h" for h in range(24)]
ALL_DEPARTMENTS = "All departments"

# Global reorder-rate-by-department ranking never changes with the slicer -- it is the
# reference chart the selected department is highlighted against.
GLOBAL_DEPT_REORDER = (
    op_full.groupby('department', observed=True)['reordered'].mean().mul(100).reset_index()
)
GLOBAL_DEPT_REORDER.columns = ['department', 'reorder_rate']
GLOBAL_DEPT_REORDER = GLOBAL_DEPT_REORDER.sort_values('reorder_rate', ascending=False).head(7)
GLOBAL_DEPT_REORDER = GLOBAL_DEPT_REORDER.sort_values('reorder_rate')

department_slicer = widgets.Dropdown(
    options=[ALL_DEPARTMENTS] + sorted(departments['department'].dropna().unique().tolist()),
    value=ALL_DEPARTMENTS,
    description="Department:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="320px"),
)

dashboard_output = widgets.Output()


def build_dashboard(selected_department: str) -> go.Figure:
    """Rebuilds the full KPI + 4-chart dashboard for the selected department slicer value."""
    if selected_department == ALL_DEPARTMENTS:
        filtered = op_full
    else:
        filtered = op_full[op_full['department'] == selected_department]

    # One row per unique order -- needed so the heatmap/day-of-week counts reflect
    # order counts, not product-line counts (a single order has many product rows).
    filtered_orders_unique = filtered.drop_duplicates(subset='order_id')[
        ['order_id', 'order_dow', 'order_hour_of_day']
    ]

    # ---------- KPI values (recomputed for the current slicer selection) ----------
    total_records = len(filtered)
    total_orders = filtered_orders_unique.shape[0]
    unique_products = filtered['product_id'].nunique()
    overall_reorder_rate = filtered['reordered'].mean() * 100 if len(filtered) else 0

    # ---------- Chart 1: Peak shopping windows heatmap (day vs hour) ----------
    heat = (
        filtered_orders_unique.groupby(['order_dow', 'order_hour_of_day'])
        .size()
        .reset_index(name='count')
    )
    heat_pivot = (
        heat.pivot(index='order_dow', columns='order_hour_of_day', values='count')
        .reindex(index=range(7), columns=range(24), fill_value=0)
    )

    # ---------- Chart 2: Reorder rate by department (global ranking, selection highlighted) ----------
    dept_reorder = GLOBAL_DEPT_REORDER.copy()
    bar_colors = [
        "#E74C3C" if (selected_department != ALL_DEPARTMENTS and d == selected_department) else "#1F9E89"
        for d in dept_reorder['department']
    ]

    # ---------- Chart 3: Orders by day of week (filtered) ----------
    dow_counts = (
        filtered_orders_unique['order_dow'].value_counts()
        .reindex(range(7), fill_value=0)
        .sort_index()
    )
    dow_counts.index = DOW_LABELS

    # ---------- Chart 4: Top 10 most popular products (filtered) ----------
    if len(filtered):
        top10 = (
            filtered.groupby('product_name', observed=True).size().reset_index(name='order_count')
            .sort_values('order_count', ascending=False).head(10).sort_values('order_count')
        )
    else:
        top10 = pd.DataFrame({"product_name": [], "order_count": []})

    # ---------- Combined figure: 4 KPI cards + 2x2 charts ----------
    fig = make_subplots(
        rows=3, cols=4,
        row_heights=[0.16, 0.42, 0.42],
        specs=[
            [{"type": "indicator"}, {"type": "indicator"}, {"type": "indicator"}, {"type": "indicator"}],
            [{"type": "heatmap", "colspan": 2}, None, {"type": "bar", "colspan": 2}, None],
            [{"type": "bar", "colspan": 2}, None, {"type": "bar", "colspan": 2}, None],
        ],
        vertical_spacing=0.11, horizontal_spacing=0.08,
        subplot_titles=("", "", "", "",
                         "Peak shopping windows (day vs. hour)", "Reorder rate by department",
                         "Orders by day of week", "Top 10 most popular products"),
    )

    kpis = [
        (f"{total_records/1e6:.2f}M" if total_records >= 1e6 else f"{total_records:,}", "RECORDS IN SELECTION", "#1F3B4D"),
        (f"{total_orders/1000:.1f}k" if total_orders >= 1000 else f"{total_orders:,}", "ORDERS IN SELECTION", "#0B8A6B"),
        (f"{unique_products/1000:.1f}k" if unique_products >= 1000 else f"{unique_products:,}", "UNIQUE PRODUCTS", "#C0532D"),
        (f"{overall_reorder_rate:.1f}%", "REORDER RATE", "#2E5FA3"),
    ]
    for i, (value, label, color) in enumerate(kpis):
        x0 = i / 4
        fig.add_annotation(x=x0 + 0.125, y=0.965, xref="paper", yref="paper",
                            text=f"<b>{value}</b>", showarrow=False,
                            font=dict(size=28, color=color), align="center")
        fig.add_annotation(x=x0 + 0.125, y=0.895, xref="paper", yref="paper",
                            text=label, showarrow=False,
                            font=dict(size=11, color="#666666"), align="center")

    fig.add_trace(go.Heatmap(z=heat_pivot.values, x=HOUR_LABELS, y=DOW_LABELS,
                              colorscale="YlGnBu", showscale=False,
                              hovertemplate="Day: %{y}<br>Hour: %{x}<br>Orders: %{z}<extra></extra>"),
                  row=2, col=1)

    fig.add_trace(go.Bar(x=dept_reorder['reorder_rate'], y=dept_reorder['department'],
                          orientation='h', marker_color=bar_colors,
                          text=dept_reorder['reorder_rate'].round(1).astype(str) + '%',
                          textposition='outside',
                          hovertemplate="%{y}: %{x:.1f}%<extra></extra>"),
                  row=2, col=3)

    fig.add_trace(go.Bar(x=dow_counts.index, y=dow_counts.values, marker_color="#4C72B0",
                          hovertemplate="%{x}: %{y:,} orders<extra></extra>"),
                  row=3, col=1)

    fig.add_trace(go.Bar(x=top10['order_count'], y=top10['product_name'],
                          orientation='h', marker_color="#8A5A12",
                          hovertemplate="%{y}: %{x:,} orders<extra></extra>"),
                  row=3, col=3)

    title_suffix = "All departments" if selected_department == ALL_DEPARTMENTS else selected_department
    fig.update_layout(
        height=980, width=1200,
        title=dict(text=f"<b>Instacart market basket analytics dashboard</b> — {title_suffix}",
                   x=0.5, xanchor="center", font=dict(size=22, color="#1F3B4D")),
        showlegend=False, plot_bgcolor="white", paper_bgcolor="white",
        margin=dict(t=140, l=60, r=40, b=40),
    )
    fig.update_xaxes(title_text="Hour of day", row=2, col=1)
    fig.update_yaxes(title_text="Day of week", row=2, col=1)
    fig.update_xaxes(title_text="Reorder rate (%)", row=2, col=3)
    fig.update_xaxes(title_text="Day of week", row=3, col=1)
    fig.update_yaxes(title_text="Number of orders", row=3, col=1)
    fig.update_xaxes(title_text="Number of orders", row=3, col=3)

    return fig


def render_current_selection(*_):
    """Clears and redraws the dashboard inside dashboard_output for the current slicer value."""
    with dashboard_output:
        dashboard_output.clear_output(wait=True)
        build_dashboard(department_slicer.value).show()


department_slicer.observe(render_current_selection, names="value")

display(widgets.VBox([department_slicer, dashboard_output]))
render_current_selection()  # initial render for the default "All departments" view
```
